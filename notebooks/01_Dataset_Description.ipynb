## 1. Introduction

Phishing is a major cybersecurity threat in which attackers attempt to deceive users by creating fraudulent websites or URLs that appear to be associated with legitimate organizations, services, or online platforms.

The main objective of a phishing attack is often to trick users into providing sensitive information such as usernames, passwords, banking credentials, payment information, or other confidential data.

Because phishing websites can closely resemble legitimate websites, manually identifying whether a URL is trustworthy can be difficult for ordinary users. Therefore, automated phishing detection techniques can play an important role in improving online security.

Machine Learning provides a way to automatically learn patterns from previously observed data. By analyzing characteristics of URLs and webpages, a Machine Learning model can learn differences between phishing and legitimate websites and use those patterns to classify previously unseen URLs.

This project focuses on developing a Machine Learning-based system for **Phishing URL Detection**.

## 3. Problem Statement

Phishing URLs can be difficult to distinguish from legitimate URLs because attackers may use techniques such as:

- Long or unusual URLs
- Suspicious domain structures
- Excessive special characters
- IP addresses instead of domain names
- Multiple subdomains
- URL redirections
- Fake webpage titles
- Suspicious external references
- Hidden webpage elements
- Other characteristics designed to imitate legitimate websites

Traditional rule-based approaches may not always generalize effectively to new phishing patterns.

Therefore, this project investigates the use of Machine Learning algorithms to learn patterns from URL and webpage characteristics and classify URLs into two categories:

**Phishing** or **Legitimate**.

## 12. Evaluation Approach

Since this is a binary classification problem, model performance will not be evaluated using accuracy alone.

The project will consider multiple evaluation metrics:

### Accuracy

Measures the proportion of correctly classified observations among all observations.

### Precision

Measures the proportion of predicted positive cases that are actually positive.

### Recall

Measures the proportion of actual positive cases correctly identified by the model.

### F1-score

Provides a combined measure based on precision and recall.

### ROC-AUC

Measures the model's ability to distinguish between the two classes across different classification thresholds.

### Confusion Matrix

Provides a detailed view of correct and incorrect predictions for each class.

The exact evaluation results will be calculated after the models are trained.

The analysis will not remove records automatically.

```python
# Analyzing URL-level duplication

total_urls = len(df)

unique_urls = df["URL"].nunique()

duplicate_url_records = total_urls - unique_urls

duplicate_url_percentage = round(
    duplicate_url_records / total_urls * 100, 2
)

url_frequency = df["URL"].value_counts()

repeated_url_count = (url_frequency > 1).sum()

print("Total URL records:", total_urls)
print("Unique URLs:", unique_urls)
print("Repeated URL records:", duplicate_url_records)
print("Repeated URL record percentage:", duplicate_url_percentage, "%")
print("Number of URLs occurring more than once:", repeated_url_count)
```
