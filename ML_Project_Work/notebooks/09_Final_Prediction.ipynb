# 09 — Final Prediction and Phishing URL Classification

## Project: Phishing URL Detection Using Machine Learning

### Objective

The objective of this notebook is to use the trained machine learning models to perform final predictions on unseen URL records.

The system classifies each URL record into one of two classes:

- **0 → Phishing**
- **1 → Legitimate**

### Final Prediction Workflow

1. Load the selected-feature dataset.
2. Load the trained machine learning models.
3. Load the required feature scaler.
4. Identify the final model based on the evaluation results.
5. Prepare unseen/sample input data.
6. Generate the predicted class.
7. Calculate the prediction probability.
8. Display and interpret the final prediction.

### Input

The prediction system uses the **40 selected numerical features** obtained during feature selection.

The target variable (`label`) is not provided as an input during prediction.

### Output

For each input record, the system will display:

- Predicted class
- Predicted class name
- Prediction probability
- Final interpretation as **Phishing** or **Legitimate**

### Expected Result

This notebook demonstrates the practical use of the trained phishing URL detection model for classifying previously unseen URL records.

```python
# Import required libraries

import pandas as pd
import numpy as np
import joblib

print("FINAL PREDICTION ENVIRONMENT")
print("=" * 60)

# Load the selected-feature dataset
df = pd.read_csv("../dataset/PhiUSIIL_Selected_Features.csv")

print("Dataset loaded successfully.")
print("Dataset shape:", df.shape)
print("Number of records:", len(df))
print("Number of selected features:", len(df.columns) - 1)
print("Target column: label")

print("\nFirst 5 records:")
display(df.head())

print("\n✅ Dataset loaded successfully.")
```

```python
# Load trained models and feature scaler

logistic_model = joblib.load("../models/logistic_regression_tuned.pkl")
decision_tree_model = joblib.load("../models/decision_tree_tuned.pkl")
random_forest_model = joblib.load("../models/random_forest_tuned.pkl")

scaler = joblib.load("../models/tuning_feature_scaler.pkl")

# Store models in a dictionary
models = {
    "Logistic Regression": logistic_model,
    "Decision Tree": decision_tree_model,
    "Random Forest": random_forest_model
}

print("TRAINED MODELS LOADED")
print("=" * 60)

for model_name, model in models.items():
    print(f"✓ {model_name}: {type(model).__name__}")

print("\n✓ Feature scaler loaded:", type(scaler).__name__)
print("✓ Total models loaded:", len(models))
```

```python
# Separate input features and target variable

X = df.drop(columns=["label"])
y = df["label"]

print("FINAL PREDICTION DATA PREPARATION")
print("=" * 60)

print("Input feature shape:", X.shape)
print("Target shape:", y.shape)

print("\nNumber of input features:", X.shape[1])
print("Target classes:", sorted(y.unique()))

print("\nFeature verification:")
print("Missing values:", X.isnull().sum().sum())
print("Duplicate feature names:", X.columns.duplicated().sum())

print("\n✅ Prediction features prepared successfully.")
```

```python
# Select sample records for final prediction demonstration

# Use records from the dataset only for demonstration.
# The label is kept separately so that we can compare
# the model prediction with the actual class afterward.

sample_size = 10

sample_data = df.sample(
    n=sample_size,
    random_state=100
)

X_sample = sample_data.drop(columns=["label"])
y_actual = sample_data["label"]

print("SAMPLE DATA PREPARED")
print("=" * 60)

print("Number of sample records:", len(X_sample))
print("Number of input features:", X_sample.shape[1])

print("\nActual classes of selected records:")
print(y_actual.to_numpy())

print("\nInput feature data:")
display(X_sample)

print("\n✅ Sample prediction data prepared.")
```

```python
# Generate predictions for the sample records using all trained models

predictions = {}
probabilities = {}

# Scale sample data for models trained with StandardScaler
X_sample_scaled = scaler.transform(X_sample)

for model_name, model in models.items():

    if model_name == "Random Forest":
        # Random Forest was trained using unscaled features
        X_input = X_sample
    else:
        # Logistic Regression and Decision Tree use scaled features
        X_input = X_sample_scaled

    # Predicted class
    y_pred = model.predict(X_input)

    # Prediction probability
    y_prob = model.predict_proba(X_input)

    predictions[model_name] = y_pred

    # Probability of the predicted class
    probabilities[model_name] = np.max(y_prob, axis=1)

print("FINAL PREDICTIONS GENERATED")
print("=" * 60)

for model_name in models:
    print(f"✓ {model_name}: {len(predictions[model_name])} predictions")

print("\n✅ Predictions generated successfully.")
```

```python
# Create a final prediction results table

prediction_results = pd.DataFrame({
    "Record": range(1, len(X_sample) + 1),
    "Actual Class": y_actual.to_numpy(),
    "Logistic Regression": predictions["Logistic Regression"],
    "Decision Tree": predictions["Decision Tree"],
    "Random Forest": predictions["Random Forest"]
})

# Convert numerical labels into readable class names
class_mapping = {
    0: "Phishing",
    1: "Legitimate"
}

prediction_results["Actual Class"] = (
    prediction_results["Actual Class"].map(class_mapping)
)

for model_name in models:
    prediction_results[model_name] = (
        prediction_results[model_name].map(class_mapping)
    )

print("FINAL PREDICTION RESULTS")
print("=" * 60)

display(prediction_results)

print("\nClass Mapping:")
print("0 → Phishing")
print("1 → Legitimate")

print("\n✅ Final prediction table generated.")
```

```python
# Display prediction probabilities for each model

probability_results = pd.DataFrame({
    "Record": range(1, len(X_sample) + 1),
    "Actual Class": y_actual.map(class_mapping).to_numpy(),
    "Logistic Regression Probability": probabilities["Logistic Regression"],
    "Decision Tree Probability": probabilities["Decision Tree"],
    "Random Forest Probability": probabilities["Random Forest"]
})

print("PREDICTION PROBABILITIES")
print("=" * 60)

display(probability_results.round(4))

print("\nProbability represents the model's confidence in its predicted class.")
print("Values closer to 1.0000 indicate higher prediction confidence.")

print("\n✅ Prediction probabilities generated successfully.")
```

```python
# Compare model predictions with the actual class

comparison_results = pd.DataFrame({
    "Actual": y_actual.map(class_mapping).to_numpy(),
    "Logistic Regression": predictions["Logistic Regression"].astype(int),
    "Decision Tree": predictions["Decision Tree"].astype(int),
    "Random Forest": predictions["Random Forest"].astype(int)
})

# Calculate whether each model predicted correctly
for model_name in models:
    comparison_results[f"{model_name} Correct"] = (
        comparison_results["Actual"].map(
            {"Phishing": 0, "Legitimate": 1}
        ) == comparison_results[model_name]
    )

print("PREDICTION VERIFICATION")
print("=" * 60)

display(comparison_results)

print("\nCorrect Predictions:")
for model_name in models:
    correct = comparison_results[f"{model_name} Correct"].sum()
    total = len(comparison_results)

    print(f"{model_name}: {correct}/{total}")

print("\n✅ Predictions compared with actual classes successfully.")
```

```python
# Load evaluation results from Notebook 08

import os

evaluation_file = "../results/evaluation_results.csv"

print("FINAL MODEL SELECTION")
print("=" * 60)

if os.path.exists(evaluation_file):

    evaluation_df = pd.read_csv(evaluation_file)

    print("Evaluation results loaded successfully.")
    print("\nModel Comparison:")
    display(evaluation_df.round(4))

    # Select model with highest F1-score
    best_model_name = evaluation_df.loc[
        evaluation_df["F1-Score"].idxmax(),
        "Model"
    ]

    best_f1_score = evaluation_df.loc[
        evaluation_df["F1-Score"].idxmax(),
        "F1-Score"
    ]

    final_model = models[best_model_name]

    print("\n" + "=" * 60)
    print("FINAL MODEL")
    print("=" * 60)

    print("Selected Model:", best_model_name)
    print(f"F1-Score: {best_f1_score:.4f}")

    print("\nModel Type:", type(final_model).__name__)

else:

    print("Evaluation results file was not found.")
    print("\nPlease run this cell in Notebook 08 first:")
    print("evaluation_df.to_csv('../results/evaluation_results.csv', index=False)")
```

```python
# Final Model Selection and Prediction

import pandas as pd
import numpy as np
import joblib
from sklearn.metrics import f1_score

print("FINAL MODEL SELECTION AND PREDICTION")
print("=" * 60)

# ---------------------------------------------------------
# 1. Generate predictions for the sample using all models
# ---------------------------------------------------------

model_predictions = {}

for model_name, model in models.items():

    if model_name in ["Logistic Regression", "Decision Tree"]:
        X_input = scaler.transform(X_sample)
    else:
        # Random Forest uses unscaled features
        X_input = X_sample

    model_predictions[model_name] = model.predict(X_input)

# ---------------------------------------------------------
# 2. Calculate F1-score on the sample
# ---------------------------------------------------------

model_f1_scores = {}

for model_name in models:

    score = f1_score(
        y_actual,
        model_predictions[model_name]
    )

    model_f1_scores[model_name] = score

# Display sample F1-scores
f1_table = pd.DataFrame(
    list(model_f1_scores.items()),
    columns=["Model", "F1-Score"]
)

print("Sample Prediction Model Comparison")
print("=" * 60)

display(f1_table.round(4))

# ---------------------------------------------------------
# 3. Select model
# ---------------------------------------------------------

best_model_name = max(
    model_f1_scores,
    key=model_f1_scores.get
)

final_model = models[best_model_name]

print("\nSelected Final Model:", best_model_name)
print(f"Sample F1-Score: {model_f1_scores[best_model_name]:.4f}")

# ---------------------------------------------------------
# 4. Generate final prediction
# ---------------------------------------------------------

if best_model_name in ["Logistic Regression", "Decision Tree"]:
    X_final_input = scaler.transform(X_sample)
else:
    X_final_input = X_sample

final_predictions = final_model.predict(X_final_input)

final_probabilities = final_model.predict_proba(X_final_input)

prediction_confidence = np.max(
    final_probabilities,
    axis=1
)

# ---------------------------------------------------------
# 5. Create final prediction table
# ---------------------------------------------------------

class_mapping = {
    0: "Phishing",
    1: "Legitimate"
}

final_results = pd.DataFrame({
    "Record": range(1, len(X_sample) + 1),
    "Actual Class": y_actual.map(class_mapping).to_numpy(),
    "Predicted Class": pd.Series(final_predictions).map(class_mapping).to_numpy(),
    "Confidence": prediction_confidence
})

print("\nFINAL PREDICTION RESULTS")
print("=" * 60)

display(final_results.round(4))

print("\n✅ Final prediction completed successfully.")
```
